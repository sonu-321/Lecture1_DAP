```python
# 8. Airthmetic
a = 10
b = 3

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a % b)
print(a ** b)
```

```python
# 9. comparison operators - python
a = 10
b = 5

print(a == b) # Equal to
print(a != b) # Not equal to
print(a > b)  # Greater than
print(a < b)  # Less than
print(a >= b) # Greater than or equal to
print(a <= b) # Less than or equal to
```

```python
# 10. Assignment operators
x = 10
print(x)   # Assign 10 to x

x += 5     # Add 5 to x: x = x+5
print(x)

x -= 3     # Multiply x by 2: x = x * 2
print(x)

x *= 2     # Divide
print(x)

x /= 4
print(x)

x //= 2
print(x)

x %= 3    # reminder
print(x)

x **= 2  # power
print(x)
```
